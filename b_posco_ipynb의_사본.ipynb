# ch4. dataframe

```python
import pandas as pd
#데이터프레임 생성
data = {'이름':['홍길동','이순신','강감찬','임꺽정','이성계'],'출생년도':[1980,1986,1990,1985,1988],'점수':[1.5,1.7,3.6,2.4,2.9]}
df =pd.DataFrame(data)
df

```

```python

#데이터프레임 열이름 변경하기
df = df.rename({'출생년도':'출생'}, axis='columns')
df
```

```python

#데이터프레임 열 데이터 가져오기
df[['이름','출생']]
df.이름
```

```python

#데이터프레임 행 데이터 가져오기
df[1:3]

```

```python
# loc[]과 iloc[]
df.loc[0:1,['이름','출생']] # 행열 이름으로 범위지정, 이름이 없으면 인덱스로 지정
df.iloc[0:2,0:2] # 인덱스 번호로 선택

```

```python
# 열추가와 삭제
df['보너스']=df['점수']*5 #보너스 열 추가
df
```

```python
df['지역']=['서울','서울','부산','대구','인천'] # 지역 열 추가
df
```

```python
del df['보너스'] # 보너스 열 삭제
df
```

```python
df.loc[5] = ['김순신', 1980,3.3,'광주'] #행추가
df

```
